```python
# Create a spark dataframe and write as a delta table
print("Starting Delta table creation")

data = [("Robert", "Baratheon", "Baratheon", "Storms End", 48),
        ("Eddard", "Stark", "Stark", "Winterfell", 46),
        ("Jamie", "Lannister", "Lannister", "Casterly Rock", 29)
        ]
schema = StructType([
    StructField("firstname", StringType(), True),
    StructField("lastname", StringType(), True),
    StructField("house", StringType(), True),
    StructField("location", StringType(), True),
    StructField("age", IntegerType(), True)
])

sample_dataframe = spark.createDataFrame(data=data, schema=schema)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("WriteCSV").getOrCreate()

data = [("Alice", 1), ("Bob", 2), ("Charlie", 3)]
columns = ["Name", "ID"]
df = spark.createDataFrame(data, columns)

output_path = "/tmp/my_output_data.csv" # Adjust this path for your environment

try:
    df.write.mode("overwrite").option("header", "true").csv(output_path)
    print(f"DataFrame successfully written to {output_path}")
except Exception as e:
    print(f"Error writing DataFrame to CSV: {e}")

spark.stop()
```
